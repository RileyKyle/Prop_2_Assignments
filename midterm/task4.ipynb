# Numerical

```python
Pi = 200_000
Ti = 200
Mi = .03
ui = Mi * np.sqrt(1.4*287*Ti)
Cf = 0.08
Tw = 3500

L_tube = 1
axial_steps = 100_000

Ai = 0.05
Ae = 0.05

xlist = np.linspace(0,L_tube,axial_steps+1)
Af = lambda x: Ai#(Ae - Ai) * x + Ai

[uout,Tout,Pout], M, x_break, A_break, Thrust = quasi_1D_flow(inlet = [ui,Ti,Pi], Af = Af,N = axial_steps,Cf=Cf, L = L_tube, Tw = Tw)
```

```python
gamma = 1.4
Tt = Tout*(1+(gamma-1)/2*M**2)
Pt = Pout*(1+(gamma-1)/2*M**2)**(gamma/(gamma-1))

rhoi = Pi/(R*Ti)
m_dot = rhoi * ui * Ai
Qnet = m_dot* (Cp*(Tout[-1]-Ti)+(uout[-1]**2-ui**2)/2)


print(f'exit Mach =             {M[-1]}')
print(f'exit Velocity =         {uout[-1]} m/s\n')

print(f'exit Temp =             {Tout[-1]} k')
print(f'exit Total Temp =       {Tt[-1]} k\n')

print(f'exit Pressure =         {Pout[-1]} Pa')
print(f'exit Total Pressure =   {Pt[-1]} Pa\n')

# print(f'exit end area = {A_break}')
print(f'{Thrust =                } N')
print(f'{Qnet =                  } W')
```
