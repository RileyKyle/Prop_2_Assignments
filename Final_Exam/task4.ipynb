```python
import numpy as np
```

```python
gamma = 1.4
Gamma = (2/(gamma+1))**((gamma+1)/(2*(gamma-1)))*np.sqrt(gamma)
R = 287
Cp = 1004.5
```

# On Design

```python
pi_cR = 20
M0R = 2
T0R = 300
Tt4R = 1500
P0R = 101325
P9R = P0R
```

```python
u0R = M0R * np.sqrt(gamma*R*T0R)
u0R
```

```python
tau_rR = 1+(gamma-1)/2*M0R**2
pi_rR = (1+(gamma-1)/2*M0R**2)**(gamma/(gamma-1))

pi_rR
```

```python
pi_cR = pi_cR
tau_cR = pi_cR**((gamma-1)/gamma)
tau_cR
```

```python
tau_lambdaR = Tt4R/T0R
```

```python
tau_bR = tau_lambdaR/(tau_rR* tau_cR)
tau_bR
```

```python
tau_tR = 1 - (tau_rR) / ( tau_lambdaR) * (tau_cR-1)
pi_tR = tau_tR**(gamma/(gamma-1))

pi_tR
```

```python
pt9_p9_R = pi_rR * pi_cR * pi_tR
M9R = (((pt9_p9_R)**((gamma-1)/gamma)-1)/((gamma-1)/2))**.5

Tt9R = T0R * tau_rR * tau_cR * tau_bR * tau_tR
T9R = Tt9R/(1+(gamma-1)/2*M9R**2)

u9R = M9R * np.sqrt(gamma*R*T9R)
F_R = 50000
m_dot_air_R = F_R / (u9R - u0R)

A4R = (m_dot_air_R) * np.sqrt(R)/(Gamma * (P0R*pi_rR*pi_cR))*np.sqrt(tau_lambdaR*T0R)
```

# Off Design

```python
M0 = 1
T0 = T0R
P0 = P0R
P9 = P0

tau_t = tau_tR
Tt4 = 1500
pi_t = pi_tR
A4 = A4R
tau_lambda = Tt4/T0
```

```python
tau_r = 1+(gamma-1)/2*M0**2
pi_r = (1+(gamma-1)/2*M0**2)**(gamma/(gamma-1))

pi_r
```

```python
tau_c = 1+(tau_lambda/tau_r)*(1-tau_t)
tau_b = tau_lambda/(tau_r * tau_c)
pi_c = tau_c**(gamma/(gamma-1))

m_dot_air = Gamma / np.sqrt(R) * A4 * (P0*pi_r*pi_c)*1/(np.sqrt(tau_lambda*T0))
```

```python
u0 = M0 * np.sqrt(gamma*R*T0)

pt9_p9 = pi_r * pi_c * pi_t
M9 = (((pt9_p9)**((gamma-1)/gamma)-1)/((gamma-1)/2))**.5

Tt9 = T0R * tau_r* tau_c * tau_b* tau_t
T9 = Tt9/(1+(gamma-1)/2*M9**2)

u9 = M9 * np.sqrt(gamma*R*T9)
F = (m_dot_air) * (u9-u0)
```
