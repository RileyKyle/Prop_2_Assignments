```python
Time_left_years = 50 - 16 - 4 # year
time = Time_left_years * 365 * 24 * 60 * 60 # sec


mile2m = 1_609.344
AU = 93_000_000 * mile2m # m

distance = 100 * AU # m

Isp = 3000 # sec
alpha_prime = 200 # watt/kg
eta_a = 0.50

m_L = 2_000_000 # kg
```

```python
g_0 = 9.81
C = Isp * g_0
dV = distance / time


print(C)
alpha = dV / C


beta = alpha * (alpha/(2*(np.exp(alpha) - 1) - alpha))**(1/2)

print(alpha)
print(beta)

tau_b = (dV/beta)**2/(2*alpha_prime*eta_a)
tau_b_day = tau_b/3600/24

m_0 = m_L / (1 - (1 - np.exp(-alpha))*(1+(beta/alpha)**2))

m_p = (1 - np.exp(-dV /(C)))*m_0

m_e =  m_0 * (m_p / m_0 /(alpha_prime*eta_a*tau_b)*C**2/2)

m_p_dot = m_p / tau_b

Thrust = m_p_dot * C


print(f'{tau_b_day = } day')
print(f'{m_0 = } kg')
print(f'{m_p = } kg')
print(f'{m_p_dot = } kg/s')
print(f'{m_e = } kg')

print(f'{Thrust = } N')
```
